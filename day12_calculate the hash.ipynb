Task

Given an integer, n, and n space-separated integers as input, create a tuple,
t, of those n integers. Then compute and print the result of hash(t).

Note: hash() is one of the functions in the __ builtins __ module, so it

need not be imported.

Input Format

The first line contains an integer, n, denoting the number of elements in the

tuple.

The second line contains n space-separated integers describing the
elements in tuple t.

Output Format

Print the result of hash(t).

```python
import hashlib
n=int(input())
t=[]
for i in range(1,n+1):
    t.append(i)
T=tuple(t)
print(hash(T))
```
