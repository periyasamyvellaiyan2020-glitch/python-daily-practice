Given the participants' score sheet for your University Sports Day, you are required to find the runner-up
score. You are given n scores. Store them in a list and find the score of the runner-up.

Input Format

The first line contains n. The second line contains an array A[] of n integers each separated by a

space.

Constraints

. 2≤n≤ 10

. -100 ≤ A[i] ≤ 100

Output Format

Print the runner-up score.

Sample Input O

5
23665

Sample Output O

5

```python
n = int(input())
arr = list(map(int, input().split())) 
filtered=[]  # Creating an array to avoid the duplicate numbers
for i in arr:
    if i not in filtered:   # Filtering the duplicate
        filtered.append(i)
filtered.sort(reverse=True)    # Arranging the arr from larger to smaller
print(filtered[1])   # Taking the second which is the second maximum
```
