There are 5 students in this class whose names and grades are assembled to build the following list:

python students = [['Harry', 37.21], ['Berry', 37.21], ['Tina', 37.2],['Akriti', 41], ['Harsh', 39]]

The lowest grade of 37.2 belongs to Tina. The second lowest grade of 37.21 belongs to both Harry
and Berry, so we order their names alphabetically and print each name on a new line.

```python
import numpy as np
import pandas as pd
students=[]
for _ in range(int(input())):
    name =input()
    score = float(input())
    students.append([name,score])
students=pd.DataFrame(students,columns=['Name','Marks'])
Marks=students['Marks']
unique=[]
for i in Marks:
    if i not in unique:
        unique.append(i)
unique.sort(reverse=True)
second_low=unique[1]
for i in students.loc[Marks==second_low]['Name'].tolist():
    print(i)
```
